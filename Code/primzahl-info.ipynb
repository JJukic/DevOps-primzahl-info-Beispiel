```python
def ist_primzahl(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

def naechste_primzahl(n):
    n += 1
    while not ist_primzahl(n):
        n += 1
    return n

def primzahl_checker():
    try:
        eingabe = int(input("Gib eine ganze Zahl ein: "))
        if ist_primzahl(eingabe):
            print(f"{eingabe} ist eine Primzahl! 🧠🔢")
        else:
            np = naechste_primzahl(eingabe)
            print(f"{eingabe} ist keine Primzahl. Die nächste Primzahl ist {np}. 🚀")
    except ValueError:
        print("Das war keine gültige ganze Zahl.")

if __name__ == "__main__":
    primzahl_checker()
```
